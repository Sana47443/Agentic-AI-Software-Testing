```python
from pathlib import Path
import sys, pandas as pd
sys.path.insert(0, str(Path.cwd().parent))
from sentinelqa.test_case_designer import TestCaseDesigner
```

```python
feature = "As a user, I should be able to reset my password by entering my email address and receiving a reset link."
designer = TestCaseDesigner()
suite = designer.design(feature)
suite
```

```python
rows = []
for case in suite.test_cases:
    row = {"name": case.name, "category": case.category, "technique": case.technique, "requires_human_review": case.requires_human_review}
    row.update({f"input.{k}": v for k, v in case.input.items()})
    row.update({f"expected.{k}": v for k, v in case.expected.items()})
    rows.append(row)
df = pd.DataFrame(rows)
df
```

```python
df.to_csv("test_cases_output.csv", index=False)
print("Test cases saved to test_cases_output.csv")
```
